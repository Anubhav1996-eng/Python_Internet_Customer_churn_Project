# Data Cleaning

```python
#  Check for missing values in each column.
data.isnull().sum()
```

```python
#  Find the total number of missing values.
Missing_value=data.isnull().sum().sum()
Missing_value
```

```python
# Handle missing values appropriately.
# Gender and InternetService is a categorial columns for handling missing value we can use mode.

```

```python
# Gender
data["Gender"]=data["Gender"].fillna(data["Gender"].mode()[0])
```

```python
# InternetService
data["InternetService"]= data["InternetService"].fillna(data["Gender"].mode()[0])
```

```python
# # TotalCharges is a int for handling to missing value 
# firstly we can check any outlier is present or not
# if outlier is present so we can use median and or not present so we can use mean .
```

```python
# identified the outliers TotalCharges 
Q1=data["TotalCharges"].quantile(.25)
Q3=data["TotalCharges"].quantile(.75)

IQR = Q3-Q1

Lower_limit= Q1 -1.5*IQR
Upper_limit= Q3 + 1.5*IQR

Outliers=data[(data["TotalCharges"]< Lower_limit)|(data["TotalCharges"]> Upper_limit)]

Outliers
```

```python
# Outliers are present so we can use median for handling the missing value in Totalcharges
```

```python
data["TotalCharges"]=data["TotalCharges"].fillna(data["TotalCharges"].median())
```

```python
# At last we can checking any missing value are left or not 
data.isnull().sum()
```

```python
 # Check for duplicate records.
data.duplicated().sum()
```

```python
 # Remove duplicate records.
data=data.drop_duplicates()
```

```python
# Create a new column called Churn_Flag where Yes = 1 and No = 0.
data["Churn_Flag"]=data["Churn"].map({"Yes":1,"No":0})
data[["Churn","Churn_Flag"]].head()
```

```python
# Categorize customers based on monthly charges as Low, Medium, and High.
data["MonthlyCharges_Group"] = pd.cut(
    data["MonthlyCharges"],
    bins=[0, 40, 80, float("inf")],
    labels=["Low", "Medium", "High"]
)

print(data[["MonthlyCharges", "MonthlyCharges_Group"]].head())
```
